[expression for item in iterable(set,dict, lists) if condition]
expression - value to store in a new list
item - element from iterable
condition - optional filter

# Reduce

reduce is an inbiuilt Python function (functools module). It applies function cuumulatively to the items of an iterable, reducing the iterable to a single value

it reduces lists of values into one result by repeatedly doing and applying the same function

Why reduce=
1. Aggregate data using sun, product, maximum, minimum or any cummulatice comparison
2. Avoid writing manual loops for accumulations
3. Works very wll with lambda function

Syntax:

from functool import reduce
reduce(function, terable[,initializer])

function - mandatory - takes two arguments and returns a value
iterable - mandatory - collection (list, tuple, etc)
initializer - optional - initial value to start the reduction. if the number is provided hten computation starts with it

```python
# sum of all the numbers
from functools import reduce
nums = [1,2,4,5,6]
total = reduce(lambda num1, num2: num1 + num2, nums)
total
```

```python
total = reduce(lambda num1, num2: num1 + num2, nums, 10) # initializer is 10
total
```

```python
# max of all the numbers
nums =[55,78,34,23]
max_value = reduce(lambda num1, num2: num1 if num1 > num2 else num2, nums)
max_value
```

```python
max_value = reduce(max, nums)
max_value
```

```python
min_value = reduce(min, nums)
min_value
```

```python
mean_value = reduce(mean, nums)
mean_value
```

```python
# iterable # sequential access
# iterator do not store the entire collection in memory, they generate elemets on the fly
# useful for large datasets or streams where you have to read a file line by line
# supports lazy evaluation
iterable = [1,2,3,4,5]
iterator = iter(iterable)

print(next(iterator))
print(next(iterator))
print(next(iterator))
print(next(iterator))
print(next(iterator))
print(next(iterator))
```

generator - yield keyword
yield is a keyword in python used iniside a function that makes it a genetor fucntion
return -> throw value outside world and exit the function code
yield -> pause the functio, throw value but also remembers the function's state for next call

generator - an iterator returned by a function containing yield

```python
list(range(1, 6)) # range is function where (Startnumber, enmding no) and ending no is exclusive in nature
```

```python
def numbers():
    for i in range (1, 6): # 1-5 as 6 is exclusive in nature
        yield i

gen = numbers()
print(next(gen)) #1
print(next(gen)) #2
print(next(gen)) #3

print(f'break')

for n in gen:
    print(n)
```

```python
def numbers():
    for i in range (1, 6): # 1-5 as 6 is exclusive in nature
        return i

gen = numbers()
print(next(gen)) #1
print(next(gen)) #2
print(next(gen)) #3

print(f'break')

for n in gen:
    print(n)
```
